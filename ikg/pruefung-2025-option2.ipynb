## Einführung
Im Seminar haben wir uns verschiedene Genome und Genom-Daten angesehen. Genome von verschiedenen Organismen können unterschiedliche Längen haben, setzen sich bestehen bei DNA jedoch immer aus Sequenzen bestehend aus:
- A wie Adenin
- T wie Tymin
- C wie Cytosin
- G wie Guanin

Ein sehr häufig genutztes Konzept beim arbeiten mit Genom-Daten sind K-mere. Ein K-mer ist eine genomische Sequenz der Länge k. Spaltet man eine längere Genom-Sequenz in K-mere auf, so überlappen die K-mere immer um k-1 Basen.
### Beispiel
Die Sequenz `ATCGAA` lässt sich in folgende K-Mere mit Länge 5 aufspalten: `ATCGA` und  `TCGAA`.

## Aufgabe 1)
(6 Punkte)

Schreiben Sie die Funktion `get_kmers(sequence, k)`. Diese gibt für den String `sequence` eine Liste  von allen K-mere der Länge k zurück. Dabei sollen mehrfach auftretende K-Mere auch mehrfach in der Liste zurückgegeben werden. 

```python
def get_kmers(sequence, k):
    r = []
    for i in range(len(sequence) -k+1):
        r.append(sequence[i:i+k])
    
    return r
```

```python
# Tests für Aufgabe 1
read = "ATCGAA"
assert type(get_kmers(read,5)) == list
assert "ATCGA" in get_kmers(read,5)
assert "TCGAA" in get_kmers(read,5)
assert get_kmers(read,2) == ["AT","TC","CG","GA", "AA"]
```

## Aufgabe 2)
(15 Punkte)

Schreiben Sie die Funktion `count_kmers(sequence_list,k)`, wobei `sequence list` eine beliebig lange Liste von DNA-Sequenzen ist und `k` die K-mer Größe. Zurückgegeben werden soll EIN Dictionary, mit der Sequenz der K-mere als Key und die Anzahl des jeweiligen K-mers in ALLEN Sequenzen als Value. Nutzen Sie gegebenen falls Ihre Funktion `get_kmers()`

```python
def count_kmers(sequence_list, k):
    d = dict()
    for sequence in sequence_list:
        kmers = get_kmers(sequence, k)
        for kmer in kmers:
            if kmer in d:
                d[kmer] += 1
            else:
                d[kmer] = 1
    return d
```

```python
# Test für Aufgabe 2
assert count_kmers(["ATCGAA"],5) == {"ATCGA":1,"TCGAA":1}
assert count_kmers(["ATCG","CGAA"],2) == {"AT":1,"TC":1,"CG":2 ,"GA":1,"AA":1}
```

## Aufgabe 4)
(15 Punkte)

Schreiben Sie die Funktion `get_low_occurance_kmers(kmer_count_dict)`, die K-mere, die am seltensten `kmer_count_dict` auftreten im  (logischerweise nur die die auch in dem Dictionary auftreten) als Liste zurück gibt. Dabei sollen keine Doppelnennungen auftreten.

```python
def get_low_occurance_kmers(kmer_count_dict):

    low_counts = []
    min_count = min(kmer_count_dict.values())
    for kmer, count in kmer_count_dict.items():
        if count == min_count:
            low_counts.append(kmer)
    return low_counts
```

```python
# Tests für Aufgabe 4
assert type(get_low_occurance_kmers({"AC":2,"CT":1,"AA":3})) == list
assert get_low_occurance_kmers({"AC":2,"CT":1,"AA":3}) == ["CT"]
assert get_low_occurance_kmers({"AC":2,"CT":1,"AA":3,"TT":1}) == ["CT","TT"]
assert get_low_occurance_kmers({"AC":30,"CT":31,"TT":30}) == ["AC","TT"]
```
