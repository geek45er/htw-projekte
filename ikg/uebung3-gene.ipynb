```python
def validate_gene(genome_sequence, gene):
    start, stop = gene

    #1
    if start < 0 or stop < 0:
        return False

    #2
    if start >= stop:
        return False

    #3
    if stop > len(genome_sequence):
        return False

    genome_sequence = genome_sequence.upper()
    gene_seq = genome_sequence[start:stop]

    #4
    if not gene_seq.startswith("ATG"):
        return False

    #5
    if not (gene_seq.endswith("TAA") or gene_seq.endswith("TAG") or gene_seq.endswith("TGA")):
        return False

    #6
    if len(gene_seq) % 3 != 0:
        return False

    return True

```

```python
# Test Bedingung 1 (2 Punkte)
genome_seq = "AAATGCGCGGCTAAGG"
assert validate_gene(genome_seq,(-3,0)) == False
assert validate_gene(genome_seq,(-6,-3)) == False
assert validate_gene(genome_seq,(0,-6)) == False
```

# Validierung von Genen

Im Semeniar haben wir uns ja bereits über Gene und Genome unterhalten. Ein Genom ist die Gesamtheit aller DNA in einem Organismus. Hat der Organismus Chromosomen wie bei uns Menschen, dann ist das Genom auch in mehrere Teile geteilt. Jedes dieser Genom Fragmente besteht in der Regel aus mehreren Genen, welche dann in ein Protein translatiert werden können. Wenn man ein Gen manuell translatieren will, dann kann man das mit Hilfe der Code Sonne machen.
![Abbildung Wikipedia Code Sonne](https://upload.wikimedia.org/wikipedia/commons/thumb/7/70/Aminoacids_table.svg/1280px-Aminoacids_table.svg.png)

Dabei wird die Gen-Sequenz von innen nach außen Abgelesen. Ganz außen stehen dann die Aminosäuren die durch jeweils 3 DNA-Basen (ein so genanntes Triplet) kodiert werden. Triplets sind nicht überlappend, das heißt wir rutschen nach dem lesen jeder Aminosäure gleich um drei DNA-Basen weiter damit ergibt sich auch, dass das Gen immer 3x so viele Basen hat, wie das Protein Aminosäuren (+ das Stopp-Codon/ Stopp-Triplet). Die Länge eines Gens ist also durch drei Teilbar.

Jedes Protein beginnt in der Translation immer mit einem Start-Codon, das durch die Aminosäure Methionin(M) und das Triplet (ATG/AUG bei RNA) codiert wird. Das die Aminosäure Methionin(M) kommt zwingend am Anfang vor, kommt aber auch häufig and anderen Stellen im Protein vor. 

Desweiteren endet jedes Gen mit einem Stopp-Codon, da erst dadurch die Translation bendet wird, aber selber keine Aminosäure darstellt. Die drei Stopp-Codons in DNA/RNA schreibweise sind TAA/UAA, TAG/UAG und TGA/UGA.

---

Diese Woche sollen Sie in der Übung die Funktion `validate_gene` schreiben. Diese Funktion bekommt als Argument `genome_sequence` einen String der DNA Sequenz enthält.

Ein Beispiel für eine `genome_sequence` wäre `"AAATGCGCGGCTAAGG"`. Beachten das Genom in `genome_sequence` kann sowohl Klein-, als auch Großschreibung enthalten!

Das Argument `gene` ist ein Tuple `(start, stop)` das sich aus Start und Stopp Position des Gens zusammen setzt. Dabei entspricht die Startposition dem 0 basierten Index des Gen begins in `genome_sequence` und das Ende dem 0 basierten Index des ersten Elements nach dem Gen. Sie können also das python - stringslicing nutzen um die Gen Sequenz zu erhalten. Für das Genom Beispiel von grade wäre das Tuple `(2,14)` ein valides Beispiel mit der Sequenz `"ATGCGCGGCTAA"`.

Die Funktion soll nun folgende Eigenschaften überprüfen und nur `True` zurückgeben, wenn alle Bedingungen eines validen Genes erfüllt sind, sonst soll `False` zurückgegeben werden.  

1. Start und Stopp sind beide >= 0
2. Start ist immer < Stopp
3. Start und Stopp sind beide im Genom enthalten also <= Länge des Genoms
4. Das Gen beginnt immer mit `ATG`
5. Das Gen endet immer mit `TAA` oder `TAG` oder `TGA`.
6. Die länge der Gensequenz (Gen, nicht Genom!) ist immer durch 3 Teilbar
